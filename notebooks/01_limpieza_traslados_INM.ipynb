### Tenemos que arreglar un desajuste documentado en regla especial 3: el rango de Coahuila en 2022 está desfasada, por lo que todos los valores en el rango deberían trasladarse por una columna a la derecha. También, en el área de la columna oficina_regional, deberíamos agregar en cada fila el valor 'Coahuila'.  

```python
# Definimos el rango
mask = master['id_fila_og'].between(4878, 5373)  

# Usamos forward-fill para llenar los nombres de lugar_origen y lugar_destino para este rango, reconociendo el desfase en que se encuentran actualmente
master.loc[mask, 'lugar_origen'] = master.loc[mask, 'lugar_origen'].ffill()
master.loc[mask, 'oficina_regional'] = master.loc[mask, 'oficina_regional'].ffill()

# Definir cuáles columnas queremos mover (todas menos id_fila_og y year)
cols_to_shift = master.columns[2:]  

# Permitir tipos mixtos
master[cols_to_shift] = master[cols_to_shift].astype(object)

# Las movemos
master.loc[mask, cols_to_shift] = master.loc[mask, cols_to_shift].shift(periods=1, axis=1)

# Agregamos "Coahuila" como valor de oficina_regional para todo el rango
master.loc[mask, cols_to_shift[0]] = 'Coahuila'
```

### Regla 2:

```python
# Rellenar id_viaje con ffill
mask_dos = master['regla'] == 'Dos'
master.loc[mask_dos, 'id_viaje'] = master.loc[mask_dos, 'id_viaje'].ffill()
```

```python
# Checar que todo se llenó
master.loc[mask_dos, ['id_fila_og', 'regla', 'id_viaje']]
master.loc[mask_dos, 'id_viaje'].isna().sum() 
```

```python
# Reglas para agregar las filas: suma (columnas demográficas), concatenar (nacionalidad), 'first' para lo demás
agg_dict = {}
for col in master.columns:
    if col == 'nacionalidad':
        agg_dict[col] = lambda x: ', '.join(x.dropna().astype(str))
    elif col in ['hombres', 'mujeres', 'niñas', 'niños']:
        agg_dict[col] = 'sum'
    elif col == 'id_viaje':
        continue  # clave para agregar todo
    else:
        agg_dict[col] = 'first'  # id_fila_og y demás columnas se quedan con su valor en la primera fila de id_viaje

agg_dos = master.loc[mask_dos].groupby('id_viaje', as_index=False).agg(agg_dict)

# Reemplazamos nuestras filas con regla 2 originales con las nuevas
master = pd.concat([master[~mask_dos], agg_dos], ignore_index=True)
```
